# Figure 2
SMR by pandemic period.

```python
import os
import matplotlib.pyplot as plt
```

```python
output_dir = '../data/figures'

# colors from original paper
NON_PANDEMIC_COLOR = '#2a7aaf'
PANDEMIC_COLOR = '#b32f1e'

# general population
REFERENCE_SMR = 1

# (SMR, 95% CI lower, upper)
PERIODS = [
    {'label': 'Pre-pandemic', 'color': NON_PANDEMIC_COLOR, 'smr': (2.18, 1.69, 2.77),
     'years': {2018: (1.42, 0.71, 2.55), 2019: (2.44, 1.84, 3.17)}},
    {'label': 'Pandemic', 'color': PANDEMIC_COLOR, 'smr': (1.97, 1.64, 2.35),
     'years': {2020: (2.00, 1.58, 2.51), 2021: (1.93, 1.43, 2.55)}},
    {'label': 'Post-pandemic', 'color': NON_PANDEMIC_COLOR, 'smr': (1.88, 1.58, 2.23),
     'years': {2022: (2.14, 1.62, 2.78), 2023: (1.97, 1.44, 2.64), 2024: (1.48, 1.00, 2.09)}},
]
```

```python
YEAR_HALF_WIDTH = 0.45


def plot_period(ax, period):
    years = list(period['years'])
    color = period['color']
    start, end = years[0] - YEAR_HALF_WIDTH, years[-1] + YEAR_HALF_WIDTH

    ax.axvspan(start, end, color=color, alpha=0.1, linewidth=0)

    smr, lower, upper = period['smr']
    ax.fill_between([start, end], lower, upper, color=color, alpha=0.25, linewidth=0)
    ax.hlines(smr, start, end, color=color, linestyle='--')

    for year, (year_smr, year_lower, year_upper) in period['years'].items():
        ax.errorbar(year, year_smr, yerr=[[year_smr - year_lower], [year_upper - year_smr]],
                    fmt='o', color=color, markeredgecolor='white', markersize=8, capsize=4)

    ax.text((start + end) / 2, 3.85, period['label'], ha='center', va='top', style='italic')
```

```python
fig, ax = plt.subplots(1, 1, figsize=(8, 5))

for period in PERIODS:
    plot_period(ax, period)

ax.axhline(REFERENCE_SMR, color='black', linestyle=':', linewidth=1)

# remove upper and right spine
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['bottom'].set_color('black')
ax.spines['left'].set_color('black')

all_years = [year for period in PERIODS for year in period['years']]
ax.set_xticks(all_years)
ax.set_xlim(all_years[0] - 0.5, all_years[-1] + 0.5)
ax.set_ylim(0, 4)
ax.yaxis.set_ticks_position('left')

ax.set_ylabel('Standardized mortality ratio (SMR)')
ax.set_xlabel('Calendar year')
ax.grid(False)

plt.tight_layout()
```

```python
os.makedirs(output_dir, exist_ok=True)
fig.savefig(os.path.join(output_dir, 'figure2_smr_pandemic.pdf'), dpi=600, bbox_inches='tight')
fig.savefig(os.path.join(output_dir, 'figure2_smr_pandemic.png'), dpi=300, bbox_inches='tight')
```
